### Add documents

````python
# [STARTER] Udaplay Project
## Part 01 - Offline RAG

# Only needed for Udacity workspace
import importlib.util
import sys

# Check if 'pysqlite3' is available before importing
if importlib.util.find_spec("pysqlite3") is not None:
    import pysqlite3
    sys.modules['sqlite3'] = sys.modules.pop('pysqlite3')

import os
import json
import chromadb
from chromadb.utils import embedding_functions

### VectorDB Instance
# Instantiate your ChromaDB Client with a persistent path
chroma_client = chromadb.PersistentClient(path="chromadb")

### Collection Setup using LOCAL Embeddings (No OpenAI Key Required!)
# This bypasses the buggy OpenAI wrapper entirely and runs 100% locally
embedding_fn = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

# Safely fetch or create the collection instance
collection = chroma_client.get_or_create_collection(
    name="udaplay",
    embedding_function=embedding_fn
)

### Add documents
data_dir = "./games"

# Verify directory exists
if not os.path.exists(data_dir):
    raise FileNotFoundError(f"The directory '{data_dir}' does not exist. Check your paths.")

print(f"Beginning indexing process from folder: '{data_dir}' using local embeddings...")

indexed_count = 0
for file_name in sorted(os.listdir(data_dir)):
    if not file_name.endswith(".json"):
        continue

    file_path = os.path.join(data_dir, file_name)
    with open(file_path, "r", encoding="utf-8") as f:
        game = json.load(f)

    # Structure data cleanly using exact keys
    content = f"[{game['Platform']}] {game['Name']} ({game['YearOfRelease']}) - {game['Description']}"

    # Extract the true string string index value (e.g. '001')
    doc_id = os.path.splitext(file_name)[0]

    # Add single record to ChromaDB
    collection.add(
        ids=[doc_id],
        documents=[content],
        metadatas=[game]
    )
    indexed_count += 1

print(f"🎉 SUCCESS: Processed and indexed {indexed_count} game files into ChromaDB local cache.")

# =========================================================================
# SANITY TEST VERIFICATION QUERY
# =========================================================================
print("\n🔍 Running verification search check...")
try:
    test_results = collection.query(
        query_texts=["Find a racing game released in 1997 for PlayStation"],
        n_results=1
    )
    print("Test Search Result Match:")
    print(json.dumps(test_results['documents'], indent=2))
except Exception as test_err:
    print(f"Verification Search Failed: {test_err}")
````
